```python
# Inspect missingness by the sender's total transaction count
person_col = 'sender_account'

# Create a copy of the dataset
df_eda = df.copy()

# Count total transactions for each sender
transaction_counts = df_eda[person_col].value_counts()
df_eda['transaction_count'] = df_eda[person_col].map(transaction_counts)

# Group senders with exactly one transaction versus senders with more than one
df_eda['transaction_group'] = np.where(
    df_eda['transaction_count'] == 1,
    '1 transaction',
    'more than 1 transaction'
 )

# Calculate missing counts and percentages across transactions in each group
missing_time_by_transaction_group = (
    df_eda.groupby('transaction_group', observed=True)['time_since_last_transaction']
      .agg(
          missing_count=lambda values: values.isna().sum(),
          total_transactions='size'
      )
 )
missing_time_by_transaction_group['missing_pct'] = (
    missing_time_by_transaction_group['missing_count']
    / missing_time_by_transaction_group['total_transactions']
    * 100
 ).round(2)

missing_time_by_transaction_group
```

Comments:
- 100% of senders with exactly one transaction have a missing value in `time_since_last_transaction`. This is expected because there is no previous transaction from which to calculate the time since the last transaction.
- 17.61% of transaction rows belonging to senders with more than one transaction have a missing value in `time_since_last_transaction`. Because these senders have transaction history, we will inspect whether the missing values occur on their first transaction or on later transactions.

```python
# For repeat senders, check whether missingness occurs on the first or a later transaction
repeat_df = df_eda[df_eda['transaction_count'] > 1].copy()
repeat_df['timestamp'] = pd.to_datetime(
    repeat_df['timestamp'],
    format='mixed'
 )
repeat_df = repeat_df.sort_values([person_col, 'timestamp', 'transaction_id'])

# Assign each repeat sender a chronological transaction number
repeat_df['transaction_number'] = (
    repeat_df.groupby(person_col).cumcount() + 1
 )
repeat_df['transaction_position'] = np.where(
    repeat_df['transaction_number'] == 1,
    'first transaction',
    'later transaction'
 )

# Calculate missingness by transaction position among repeat senders
missing_time_by_position_repeat = (
    repeat_df.groupby('transaction_position', observed=True)['time_since_last_transaction']
      .agg(
          missing_count=lambda values: values.isna().sum(),
          total_transactions='size'
      )
 )
missing_time_by_position_repeat['missing_pct'] = (
    missing_time_by_position_repeat['missing_count']
    / missing_time_by_position_repeat['total_transactions']
    * 100
 ).round(2)

missing_time_by_position_repeat
```

```python
# Explore whether missingness is related to observed variables
analysis_df = df_eda.copy()
analysis_df['time_since_last_transaction_missing'] = (
    analysis_df['time_since_last_transaction'].isna()
 )
analysis_df['amount_decile'] = pd.qcut(
    analysis_df['amount'],
    q=10,
    duplicates='drop'
 )
analysis_df['transaction_hour'] = pd.to_datetime(
    analysis_df['timestamp'],
    format='mixed'
 ).dt.hour

observed_features = [
    'transaction_count',
    'transaction_group',
    'amount_decile',
    'transaction_type',
    'merchant_category',
    'location',
    'device_used',
    'transaction_hour',
    'is_fraud'
 ]

print(
    'Overall missing percentage:',
    f"{analysis_df['time_since_last_transaction_missing'].mean() * 100:.2f}%"
 )

for feature in observed_features:
    feature_summary = (
        analysis_df.groupby(feature, observed=True)['time_since_last_transaction_missing']
          .agg(
              missing_count='sum',
              total_rows='size'
          )
     )
    feature_summary['missing_pct'] = (
        feature_summary['missing_count']
        / feature_summary['total_rows']
        * 100
     ).round(2)
    print(f'\nMissingness by {feature}:')
    display(feature_summary)
```

Comments:
- The missingness is not completely random (MCAR): missingness varies systematically with the sender's transaction count, with exactly one missing value per sender.
- Missingness is strongly associated with the target: all fraudulent transactions have a non-missing `time_since_last_transaction`, while all missing values occur on non-fraudulent transactions. This suggests a synthetic data-generation rule or target leakage and requires further investigation before modeling.

### Outliers

```python
# Detect potential outliers in numeric transaction features using the IQR rule
numeric_columns = [
    'amount',
    'time_since_last_transaction',
    'spending_deviation_score',
    'velocity_score',
    'geo_anomaly_score'
 ]

outlier_summary = []

for column in numeric_columns:
    values = df[column].dropna()
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_mask = (df[column] < lower_bound) | (df[column] > upper_bound)

    outlier_summary.append({
        'column': column,
        'q1': q1,
        'q3': q3,
        'lower_bound': lower_bound,
        'upper_bound': upper_bound,
        'outlier_count': outlier_mask.sum(),
        'outlier_pct': outlier_mask.mean() * 100
     })

outlier_summary = (
    pd.DataFrame(outlier_summary)
      .sort_values('outlier_count', ascending=False)
 )
outlier_summary['outlier_pct'] = outlier_summary['outlier_pct'].round(2)

outlier_summary
```

Comments:

## Validate `time_since_last_transaction`

The original value is compared with a value recalculated from timestamps after sorting each sender chronologically. The first transaction for each sender has no previous transaction, so its recalculated value is missing.

```python
# Recalculate time since the sender's previous transaction
time_check = df[[
    'transaction_id',
    'sender_account',
    'timestamp',
    'time_since_last_transaction'
]].copy()

time_check['timestamp'] = pd.to_datetime(
    time_check['timestamp'],
    format='mixed',
    errors='coerce'
)
time_check['original_time_since_last_transaction'] = pd.to_numeric(
    time_check['time_since_last_transaction'],
    errors='coerce'
)

# Sort transactions in chronological order for each sender
time_check = (
    time_check
    .sort_values(['sender_account', 'timestamp', 'transaction_id'])
    .reset_index(drop=True)
)

time_check['recalculated_time_since_last_transaction'] = (
    time_check.groupby('sender_account')['timestamp']
    .diff()
    .dt.total_seconds()
)

# Compare values
time_check['matches'] = (
    np.isclose(
        time_check['original_time_since_last_transaction'].fillna(-1),
        time_check['recalculated_time_since_last_transaction'].fillna(-1),
        rtol=0,
        atol=1
    )
)

comparison_summary = pd.DataFrame({
    'metric': [
        'total rows',
        'matching rows',
        'different rows',
        'original negative values',
        'recalculated negative values',
        'original missing values',
        'recalculated missing values'
    ],
    'value': [
        len(time_check),
        time_check['matches'].sum(),
        (~time_check['matches']).sum(),
        (time_check['original_time_since_last_transaction'] < 0).sum(),
        (time_check['recalculated_time_since_last_transaction'] < 0).sum(),
        time_check['original_time_since_last_transaction'].isna().sum(),
        time_check['recalculated_time_since_last_transaction'].isna().sum()
    ]
})

display(comparison_summary)

# Inspect the first mismatches
mismatches = time_check.loc[
    ~time_check['matches'],
    [
        'transaction_id',
        'sender_account',
        'timestamp',
        'original_time_since_last_transaction',
        'recalculated_time_since_last_transaction'
    ]
].copy()
mismatches['difference_seconds'] = (
    mismatches['original_time_since_last_transaction']
    - mismatches['recalculated_time_since_last_transaction']
)

display(mismatches.head(20))
mismatches.to_csv(
    os.path.join(OUTPUT_DIR, 'time_since_transaction_mismatches.csv'),
    index=False
)
```

## Data Modeling

Time-based split: The raw transactions are split chronologically before feature engineering. Historical features are calculated using only the current transaction and transactions occurring earlier in time.

```python
# Split raw transactions before calculating any history-based features
raw_model_data = df.copy()
raw_model_data['timestamp'] = pd.to_datetime(
    raw_model_data['timestamp'], format='mixed', errors='coerce'
)
raw_model_data['is_fraud'] = pd.to_numeric(
    raw_model_data['is_fraud'], errors='coerce'
)
raw_model_data = (
    raw_model_data.dropna(subset=['timestamp', 'is_fraud'])
    .sort_values(['timestamp', 'transaction_id'])
    .reset_index(drop=True)
)
raw_model_data['is_fraud'] = raw_model_data['is_fraud'].astype(int)

n_rows = len(raw_model_data)
train_end = int(n_rows * 0.70)
validation_end = int(n_rows * 0.85)

train_raw = raw_model_data.iloc[:train_end].copy()
validation_raw = raw_model_data.iloc[train_end:validation_end].copy()
test_raw = raw_model_data.iloc[validation_end:].copy()

print(train_raw.shape, validation_raw.shape, test_raw.shape)
```

```python
def add_prior_features(frame):
    frame = frame.copy()
    frame = frame.sort_values(['timestamp', 'transaction_id']).reset_index(drop=True)

    # Calendar features use only the transaction timestamp.
    frame['hour'] = frame['timestamp'].dt.hour
    frame['day_of_week'] = frame['timestamp'].dt.dayofweek
    frame['month'] = frame['timestamp'].dt.month
    frame['is_weekend'] = frame['day_of_week'].isin([5, 6]).astype(int)
    frame['is_night'] = frame['hour'].isin([0, 1, 2, 3, 4, 5]).astype(int)

    # Prior sender statistics. Current amount is excluded from the history.
    sender = frame.groupby('sender_account', sort=False, dropna=False)
    frame['sender_prior_count'] = sender.cumcount()
    frame['time_since_sender_transaction'] = (
        sender['timestamp'].diff().dt.total_seconds()
    )
    prior_sum = sender['amount'].cumsum() - frame['amount']
    prior_squared_sum = (
        frame['amount'].pow(2).groupby(
            frame['sender_account'], sort=False, dropna=False
        ).cumsum() - frame['amount'].pow(2)
    )
    prior_count = frame['sender_prior_count'].replace(0, np.nan)
    frame['sender_prior_mean'] = prior_sum / prior_count
    frame['sender_prior_variance'] = (
        prior_squared_sum / prior_count - frame['sender_prior_mean'].pow(2)
    ).clip(lower=0)
    frame['sender_prior_std'] = np.sqrt(frame['sender_prior_variance'])
    frame['amount_vs_sender_prior_mean'] = (
        frame['amount'] / frame['sender_prior_mean']
    ).replace([np.inf, -np.inf], np.nan)

    # Explicit "no history yet" flag, since NaN in the columns above means no prior transaction, the imputer needs this
    # flag to tell those rows apart from a real zero-valued observation.
    frame['is_first_sender_transaction'] = (
        frame['sender_prior_count'] == 0
    ).astype(int)

    # Prior receiver and sender-receiver relationship statistics.
    receiver = frame.groupby('receiver_account', sort=False, dropna=False)
    frame['receiver_prior_count'] = receiver.cumcount()
    pair = frame.groupby(
        ['sender_account', 'receiver_account'], sort=False, dropna=False
    )
    frame['sender_receiver_prior_count'] = pair.cumcount()
    frame['time_since_sender_receiver_transaction'] = (
        pair['timestamp'].diff().dt.total_seconds()
    )
    frame['is_first_sender_receiver_pair'] = (
        frame['sender_receiver_prior_count'] == 0
    ).astype(int)

    # Prior device and IP counts.
    for column, prefix in [('device_hash', 'device'), ('ip_address', 'ip')]:
        frame[f'{prefix}_prior_count'] = frame.groupby(
            column, sort=False, dropna=False
        ).cumcount()
        frame[f'is_new_{prefix}'] = (
            frame[f'{prefix}_prior_count'] == 0
        ).astype(int)

    frame['log_amount'] = np.log1p(frame['amount'].clip(lower=0))
    frame['amount_round_number_flag'] = (frame['amount'] % 100 == 0).astype(int)
    frame['rapid_transaction_flag'] = (
        frame['time_since_sender_transaction'] < 60
    ).fillna(False).astype(int)
    return frame
```

```python
# Calculate features after splitting
feature_input = pd.concat([
    train_raw.assign(_split='train'),
    validation_raw.assign(_split='validation'),
    test_raw.assign(_split='test')
], ignore_index=True)
featured_data = add_prior_features(feature_input)

train_features = featured_data[featured_data['_split'] == 'train'].copy()
validation_features = featured_data[featured_data['_split'] == 'validation'].copy()
test_features = featured_data[featured_data['_split'] == 'test'].copy()

NON_MODEL_COLUMNS = [
    'is_fraud', '_split', 'timestamp', 'transaction_id',
    'sender_account', 'receiver_account', 'device_hash',
    'ip_address', 'fraud_type', 'time_since_last_transaction',
    'transaction_count', 'transaction_group'
]
feature_columns = [
    c for c in train_features.columns if c not in NON_MODEL_COLUMNS
]

X_train = train_features[feature_columns].copy()
y_train = train_features['is_fraud'].copy()
X_validation = validation_features[feature_columns].copy()
y_validation = validation_features['is_fraud'].copy()
X_test = test_features[feature_columns].copy()
y_test = test_features['is_fraud'].copy()

print(f'Features: {len(feature_columns)}')
print('Train fraud rate:', y_train.mean())
print('Validation fraud rate:', y_validation.mean())
print('Test fraud rate:', y_test.mean())
```

## Class-Weighted Logistic Regression Baseline

This baseline uses `class_weight='balanced'`

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score, classification_report,
    confusion_matrix, precision_recall_curve, roc_auc_score
)

numeric_features = X_train.select_dtypes(include='number').columns.tolist()
categorical_features = X_train.select_dtypes(
    include=['object', 'category', 'string']
).columns.tolist()

preprocessor = ColumnTransformer([
    ('numeric', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), numeric_features),
    ('categorical', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=True))
    ]), categorical_features)
])

baseline_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        class_weight='balanced', max_iter=1000,
        solver='liblinear', random_state=42
    ))
])

baseline_model.fit(X_train, y_train)
validation_probability = baseline_model.predict_proba(X_validation)[:, 1]

print('Validation ROC-AUC:', roc_auc_score(y_validation, validation_probability))
print('Validation PR-AUC:', average_precision_score(y_validation, validation_probability))
```

```python
# Tune the classification threshold on validation data using F1
precision, recall, thresholds = precision_recall_curve(
    y_validation, validation_probability
)
f1_scores = 2 * precision[:-1] * recall[:-1] / (
    precision[:-1] + recall[:-1] + 1e-12
)
best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]
validation_prediction = (
    validation_probability >= best_threshold
).astype(int)

print(f'Best validation threshold: {best_threshold:.4f}')
print(classification_report(
    y_validation, validation_prediction,
    target_names=['non-fraud', 'fraud'], zero_division=0
))
print('Confusion matrix:')
print(confusion_matrix(y_validation, validation_prediction))
```

```python
# Evaluate the selected threshold once on the untouched test set
test_probability = baseline_model.predict_proba(X_test)[:, 1]
test_prediction = (test_probability >= best_threshold).astype(int)

test_report = classification_report(
    y_test, test_prediction, output_dict=True, zero_division=0
)
test_cm = confusion_matrix(y_test, test_prediction)
tn, fp, fn, tp = test_cm.ravel()

test_precision = test_report['1']['precision']
test_recall = test_report['1']['recall']
test_f1 = test_report['1']['f1-score']
test_f2 = (5 * test_precision * test_recall) / (
    4 * test_precision + test_recall + 1e-12
)
test_specificity = tn / (tn + fp + 1e-12)
test_fpr = fp / (fp + tn + 1e-12)
test_fnr = fn / (fn + tp + 1e-12)

print('Test ROC-AUC:', roc_auc_score(y_test, test_probability))
print('Test PR-AUC:', average_precision_score(y_test, test_probability))
print(f'Test threshold: {best_threshold:.4f}')
print(classification_report(
    y_test, test_prediction,
    target_names=['non-fraud', 'fraud'], zero_division=0
))
print('Test confusion matrix:')
print(test_cm)
```

## Imbalance and Model Comparison

All resampling is applied only to the training matrix. Validation and test data retain their natural fraud rate. The test set is evaluated only after selecting the approach and threshold using validation data.

```python
# Fit preprocessing once using training data only
X_train_encoded = preprocessor.fit_transform(X_train)
X_validation_encoded = preprocessor.transform(X_validation)
X_test_encoded = preprocessor.transform(X_test)

print("Encoded train shape:", X_train_encoded.shape)
print("Encoded validation shape:", X_validation_encoded.shape)
print("Encoded test shape:", X_test_encoded.shape)
```

```python
from imblearn.over_sampling import RandomOverSampler, SMOTE
from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    average_precision_score, balanced_accuracy_score, classification_report,
    confusion_matrix, f1_score, precision_score, recall_score, roc_auc_score
)
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

def select_f1_threshold(labels, probabilities):
    precision_values, recall_values, threshold_values = precision_recall_curve(
        labels, probabilities
    )
    f1_values = (
        2 * precision_values[:-1] * recall_values[:-1]
        / (precision_values[:-1] + recall_values[:-1] + 1e-12)
    )
    index = int(np.nanargmax(f1_values))
    return float(threshold_values[index])

def calculate_metrics(name, strategy, labels, probabilities, threshold):
    predictions = (probabilities >= threshold).astype(int)
    tn_value, fp_value, fn_value, tp_value = confusion_matrix(
        labels, predictions, labels=[0, 1]
    ).ravel()
    precision_value = precision_score(labels, predictions, zero_division=0)
    recall_value = recall_score(labels, predictions, zero_division=0)
    return {
        "model": name,
        "strategy": strategy,
        "threshold": threshold,
        "precision": precision_value,
        "recall": recall_value,
        "f1": f1_score(labels, predictions, zero_division=0),
        "f2": (5 * precision_value * recall_value) / (4 * precision_value + recall_value + 1e-12),
        "pr_auc": average_precision_score(labels, probabilities),
        "roc_auc": roc_auc_score(labels, probabilities),
        "balanced_accuracy": balanced_accuracy_score(labels, predictions),
        "specificity": tn_value / (tn_value + fp_value + 1e-12),
        "false_positive_rate": fp_value / (fp_value + tn_value + 1e-12),
        "false_negative_rate": fn_value / (fn_value + tp_value + 1e-12)
    }

def fit_and_score(model, name, strategy, train_matrix, train_labels):
    model.fit(train_matrix, train_labels)
    validation_scores = model.predict_proba(X_validation_encoded)[:, 1]
    threshold = select_f1_threshold(y_validation, validation_scores)
    metrics = calculate_metrics(
        name, strategy, y_validation, validation_scores, threshold
    )
    return model, metrics, validation_scores
```

```python
# Random oversampling: training data only; validation remains untouched
random_sampler = RandomOverSampler(sampling_strategy=0.25, random_state=42)
X_train_ros, y_train_ros = random_sampler.fit_resample(
    X_train_encoded, y_train
)
ros_model = LogisticRegression(
    max_iter=1000, solver="liblinear", random_state=42
)
ros_model, ros_metrics, ros_validation_probability = fit_and_score(
    ros_model, "Logistic Regression", "random_oversampling_0.25",
    X_train_ros, y_train_ros
)
print("Original training rows:", X_train_encoded.shape[0])
print("Resampled training rows:", X_train_ros.shape[0])
print(ros_metrics)
```

```python
# SMOTE: training data only; using a moderate minority ratio to control memory use
# Note: this is a diagnostic comparison after one-hot encoding. For a production
# experiment, use SMOTENC before one-hot encoding so categories are not interpolated.
smote_sampler = SMOTE(sampling_strategy=0.25, random_state=42, k_neighbors=5)
X_train_smote, y_train_smote = smote_sampler.fit_resample(
    X_train_encoded, y_train
)
smote_model = LogisticRegression(
    max_iter=1000, solver="liblinear", random_state=42
)
smote_model, smote_metrics, smote_validation_probability = fit_and_score(
    smote_model, "Logistic Regression", "smote_0.25",
    X_train_smote, y_train_smote
)
print("SMOTE training rows:", X_train_smote.shape[0])
print(smote_metrics)
```

```python
# Train tree-based models with class weighting or scale_pos_weight
from catboost import CatBoostClassifier

negative_count = int((y_train == 0).sum())
positive_count = int((y_train == 1).sum())
fraud_weight = negative_count / max(positive_count, 1)

tree_models = [
    (
        "Random Forest",
        "class_weight=balanced",
        RandomForestClassifier(
            n_estimators=100, max_depth=12, min_samples_leaf=5,
            class_weight="balanced", n_jobs=-1, random_state=42
        )
    ),
    (
        "XGBoost",
        "scale_pos_weight",
        XGBClassifier(
            n_estimators=150, max_depth=6, learning_rate=0.08,
            subsample=0.8, colsample_bytree=0.8,
            scale_pos_weight=fraud_weight, objective="binary:logistic",
            eval_metric="aucpr", n_jobs=-1, random_state=42
        )
    ),
    (
        "LightGBM",
        "scale_pos_weight",
        LGBMClassifier(
            n_estimators=150, learning_rate=0.08, num_leaves=31,
            subsample=0.8, colsample_bytree=0.8,
            scale_pos_weight=fraud_weight, verbosity=-1,
            n_jobs=-1, random_state=42
        )
    ),
    (
        "CatBoost",
        "class_weights",
        CatBoostClassifier(
            iterations=150, depth=8, learning_rate=0.08,
            class_weights=[1.0, fraud_weight],
            loss_function="Logloss", eval_metric="AUC:PR",
            verbose=False, random_seed=42, thread_count=-1
        )
    )
]

# Store classifiers that all expect the already-encoded feature matrix.
baseline_classifier = baseline_model.named_steps["classifier"]
model_runs = [
    ("Logistic Regression", "class_weight=balanced", baseline_classifier, validation_probability),
    ("Logistic Regression", "random_oversampling_0.25", ros_model, ros_validation_probability),
    ("Logistic Regression", "smote_0.25", smote_model, smote_validation_probability)
]
tree_validation_scores = {}

for name, strategy, model in tree_models:
    fitted_model, metrics, scores = fit_and_score(
        model, name, strategy, X_train_encoded, y_train
    )
    model_runs.append((name, strategy, fitted_model, scores))
    tree_validation_scores[name] = scores
    print(name, metrics)
```

```python
# Build the validation comparison table and choose the candidate by PR-AUC, then F1
validation_rows = []
for name, strategy, model, scores in model_runs:
    threshold = select_f1_threshold(y_validation, scores)
    validation_rows.append(
        calculate_metrics(name, strategy, y_validation, scores, threshold)
    )

validation_comparison = (
    pd.DataFrame(validation_rows)
    .sort_values(["pr_auc", "f1"], ascending=False)
    .reset_index(drop=True)
)
display(validation_comparison.round(4))
validation_comparison.to_csv(
    os.path.join(OUTPUT_DIR, "model_validation_comparison.csv"),
    index=False
)

selected_name = validation_comparison.loc[0, "model"]
selected_strategy = validation_comparison.loc[0, "strategy"]
selected_threshold = float(validation_comparison.loc[0, "threshold"])
selected_model = next(
    model for name, strategy, model, scores in model_runs
    if name == selected_name and strategy == selected_strategy
)
print("Selected candidate:", selected_name, selected_strategy)
```

```python
# Evaluate the selected candidate once on the untouched test set
selected_test_probability = selected_model.predict_proba(X_test_encoded)[:, 1]
test_metrics = calculate_metrics(
    selected_name, selected_strategy, y_test,
    selected_test_probability, selected_threshold
)
test_comparison = pd.DataFrame([{**test_metrics, "dataset": "test"}])
display(test_comparison.round(4))

# Append the final test result to the comparison artifact
metrics_table = pd.concat([
    validation_comparison.assign(dataset="validation"),
    test_comparison
], ignore_index=True)
metrics_table.to_csv(
    os.path.join(OUTPUT_DIR, "model_metrics_comparison.csv"),
    index=False
)
print("Complete comparison saved to:", os.path.join(OUTPUT_DIR, "model_metrics_comparison.csv"))
```

```python
# Save the selected model, preprocessing transformer, threshold, and feature list
selected_artifact = {
    "model": selected_model,
    "preprocessor": preprocessor,
    "feature_columns": feature_columns,
    "threshold": selected_threshold,
    "model_name": selected_name,
    "strategy": selected_strategy
}
selected_model_path = os.path.join(OUTPUT_DIR, "fraud_selected_model.joblib")
joblib.dump(selected_artifact, selected_model_path)
print("Selected model saved to:", selected_model_path)
```

```python
# Final comparison: baseline Logistic Regression vs. the selected candidate,
# whether there is a clear winner among candidates, and whether the selected
# candidate's test performance meets a business bar.
baseline_summary = {
    "model": "Logistic Regression",
    "strategy": "class_weight=balanced",
    "precision": test_precision,
    "recall": test_recall,
    "f1": test_f1,
    "f2": test_f2,
    "pr_auc": average_precision_score(y_test, test_probability),
    "roc_auc": roc_auc_score(y_test, test_probability),
}

comparison_metrics = ["precision", "recall", "f1", "f2", "pr_auc", "roc_auc"]
improvement = pd.DataFrame({
    "metric": comparison_metrics,
    "baseline_LR_test": [baseline_summary[m] for m in comparison_metrics],
    "selected_model_test": [test_metrics[m] for m in comparison_metrics],
})
improvement["absolute_gain"] = (
    improvement["selected_model_test"] - improvement["baseline_LR_test"]
)
print(f"Selected candidate: {selected_name} ({selected_strategy})")
display(improvement.round(4))

```
